# Functions: 
- sample_data
- run_affinity_propagation
- run_dbscan
- run_birch
- filter_clusters
- plot_three_clusters

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.cluster import AffinityPropagation
from sklearn.cluster import DBSCAN
from sklearn.cluster import Birch


def sample_data(dataframe, sample_size=500, random_state=42):
    """
    Take a random sample of the dataframe.
    """
    return dataframe.sample(
        n=min(sample_size, len(dataframe)),
        random_state=random_state
    )


def run_affinity_propagation(
    dataframe,
    damping=0.9,
    max_iter=300,
    convergence_iter=15,
    random_state=42
):
    model = AffinityPropagation(
        damping=damping,
        max_iter=max_iter,
        convergence_iter=convergence_iter,
        random_state=random_state
    )

    labels = model.fit_predict(dataframe)

    return model, labels


def run_dbscan(
    dataframe,
    eps=1.0,
    min_samples=5
):
    model = DBSCAN(
        eps=eps,
        min_samples=min_samples
    )

    labels = model.fit_predict(dataframe)

    return model, labels


def run_birch(
    dataframe,
    n_clusters=4
):
    model = Birch(
        n_clusters=n_clusters
    )

    labels = model.fit_predict(dataframe)

    return model, labels


def print_cluster_info(name, labels):

    counts = pd.Series(
        labels
    ).value_counts().sort_index()

    print(f"\n{name}")
    print("-" * len(name))

    for cluster, count in counts.items():

        if cluster == -1:
            print(f"Noise: {count}")
        else:
            print(
                f"Cluster {cluster}: {count}"
            )

    n_clusters = len(
        set(labels) - {-1}
    )

    print(
        f"Number of clusters: {n_clusters}"
    )
```

# Task 3 
# Affinity Propagation Clustering, DBSCAN Clustering, BIRCH Clustering

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.decomposition import PCA


# Make dataframe
df_scaled = pd.DataFrame(
    X_scaled,
    index=df.index
)


# 1. SAMPLE THE SCALED DATA
df_sample = sample_data(
    df_scaled,
    sample_size=500,
    random_state=42
)

print(
    "Original sample size:",
    len(df_sample)
)


# 2. REMOVE EXTREME OUTLIERS
isolation_forest = IsolationForest(
    contamination=0.02,
    random_state=42
)

outlier_labels = isolation_forest.fit_predict(
    df_sample
)

df_sample_clean = df_sample[
    outlier_labels == 1
].copy()

print(
    "Clean sample size:",
    len(df_sample_clean)
)

print(
    "Outliers removed:",
    len(df_sample) - len(df_sample_clean)
)


# 3. PCA
pca = PCA(
    n_components=0.9,
    random_state=42
)

X_pca = pca.fit_transform(
    df_sample_clean
)

print(
    "PCA dimensions:",
    X_pca.shape[1]
)

print(
    "Explained variance:"
)

for i, variance in enumerate(
    pca.explained_variance_ratio_
):
    print(
        f"PC{i+1}: {variance * 100:.2f}%"
    )

print(
    "PC1 + PC2:",
    f"{pca.explained_variance_ratio_[:2].sum() * 100:.2f}%"
)


# 4. RUN AFFINITY PROPAGATION
af, labels_ap = run_affinity_propagation(
    X_pca,
    damping=0.99,
    max_iter=300,
    convergence_iter=15,
    random_state=42
)


# 5. RUN DBSCAN
dbscan, labels_dbscan = run_dbscan(
    X_pca,
    eps=0.5,
    min_samples=5
)


# 6. RUN BIRCH
brc, labels_birch = run_birch(
    X_pca,
    n_clusters=4
)


# 7. PRINT CLUSTER INFORMATION
print_cluster_info(
    "Affinity Propagation",
    labels_ap
)

print_cluster_info(
    "DBSCAN",
    labels_dbscan
)

print_cluster_info(
    "BIRCH",
    labels_birch
)


# 8. PLOT
results = [
    ("Affinity Propagation", labels_ap),
    ("DBSCAN", labels_dbscan),
    ("BIRCH", labels_birch)
]


pc1_ratio = (
    pca.explained_variance_ratio_[0] * 100
)

pc2_ratio = (
    pca.explained_variance_ratio_[1] * 100
)


fig, axes = plt.subplots(
    1,
    3,
    figsize=(20, 6)
)


for ax, (name, labels) in zip(
    axes,
    results
):

    unique_labels = sorted(
        set(labels)
    )

    n_clusters = len([
        label
        for label in unique_labels
        if label != -1
    ])

    label_map = {
        label: i
        for i, label in enumerate(
            unique_labels
        )
    }

    colors = [
        label_map[label]
        for label in labels
    ]

    cmap = plt.get_cmap(
        "tab10",
        max(len(unique_labels), 1)
    )

    scatter = ax.scatter(
        X_pca[:, 0],
        X_pca[:, 1],
        c=colors,
        cmap=cmap,
        alpha=0.65,
        s=35
    )

    ax.set_xlabel(
        f"PC1 ({pc1_ratio:.1f}%)"
    )

    ax.set_ylabel(
        f"PC2 ({pc2_ratio:.1f}%)"
    )

    ax.set_title(
        f"{name} — {n_clusters} clusters"
    )

    ax.grid(
        alpha=0.2
    )

    colorbar = fig.colorbar(
        scatter,
        ax=ax,
        ticks=range(len(unique_labels)),
        pad=0.02
    )

    colorbar.set_label(
        "Cluster"
    )

    colorbar.set_ticklabels(
        unique_labels
    )


plt.tight_layout()
plt.show()
```

The PC dimensionality reduction method resulted in the same object placement in all 3 plots. Affinity Propagation, DBSCAN and BIRCH all resulted in different classification results. The dimensionality (number of components) was set to 41. The top 2 principal components (PCA 1, PCA2) were used for the visualisations. For each visualisation, PC1 explains 9.9% of the variance, and PC2 explains 6.8% of the variance. All of the plots have noise on the right side.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


df = pd.read_csv("online_shoppers_intention.csv")


# ---------------------------------------------------------
# 1. Remove the target variable from clustering
# ---------------------------------------------------------

# Revenue is an outcome, not a feature for discovering clusters.
revenue = df["Revenue"].astype(int)

df_features = df.drop(columns=["Revenue"]).copy()


# ---------------------------------------------------------
# 2. Convert boolean variable
# ---------------------------------------------------------

df_features["Weekend"] = df_features["Weekend"].astype(int)


# ---------------------------------------------------------
# 3. One-hot encode categorical variables
# ---------------------------------------------------------

categorical_columns = [
    "Month",
    "OperatingSystems",
    "Browser",
    "Region",
    "TrafficType",
    "VisitorType"
]

df_features = pd.get_dummies(
    df_features,
    columns=categorical_columns,
    dtype=float
)


# ---------------------------------------------------------
# 4. Scale ALL features
# ---------------------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(df_features)


print("Number of observations:", X_scaled.shape[0])
print("Number of features:", X_scaled.shape[1])


# ---------------------------------------------------------
# 5. PCA
# ---------------------------------------------------------

# Keep enough components to retain 90% of the variance.
pca = PCA(
    n_components=3,
    random_state=42
)

X_pca = pca.fit_transform(X_scaled)

print("PCA dimensions:", X_pca.shape[1])
print(
    "Explained variance:",
    pca.explained_variance_ratio_.sum()
)
```

```python
from sklearn.cluster import AffinityPropagation, DBSCAN, Birch
from sklearn.neighbors import NearestNeighbors
import numpy as np
import pandas as pd


# ---------------------------------------------------------
# 6. Sample data
# ---------------------------------------------------------

df_sample = df.sample(
    n=1000,
    random_state=42
)

X_sample = df_sample.drop(
    columns=["Revenue"]
).values


# ---------------------------------------------------------
# 7. Scale the data
# ---------------------------------------------------------

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_sample)


# ---------------------------------------------------------
# 8. PCA
# ---------------------------------------------------------

from sklearn.decomposition import PCA

pca = PCA(
    n_components=5
)

X_pca = pca.fit_transform(X_scaled)


# ---------------------------------------------------------
# 9. Affinity Propagation
# ---------------------------------------------------------

af = AffinityPropagation(
    damping=0.9,
    max_iter=300,
    random_state=42
)

labels_ap = af.fit_predict(X_pca)


# ---------------------------------------------------------
# 10. DBSCAN
# ---------------------------------------------------------

dbscan = DBSCAN(
    eps=1.5,
    min_samples=10
)

labels_dbscan = dbscan.fit_predict(X_pca)


# ---------------------------------------------------------
# 11. BIRCH
# ---------------------------------------------------------

birch = Birch(
    n_clusters=4
)

labels_birch = birch.fit_predict(X_pca)


# ---------------------------------------------------------
# 12. Print results
# ---------------------------------------------------------

algorithms = [
    ("Affinity Propagation", labels_ap),
    ("DBSCAN", labels_dbscan),
    ("BIRCH", labels_birch)
]

for name, labels in algorithms:

    print("\n", name)

    print(
        pd.Series(labels)
        .value_counts()
        .sort_index()
    )

    print(
        "Number of clusters:",
        len(set(labels) - {-1})
    )

    print(
        "Number of noise points:",
        sum(labels == -1)
    )
```

```python
# ---------------------------------------------------------
# 11. PCA for visualization
# ---------------------------------------------------------

# X_sample already contains all PCA dimensions.
# Only the first two are used for visualization.

pc1 = X_sample[:, 0]
pc2 = X_sample[:, 1]

pc1_ratio = pca.explained_variance_ratio_[0] * 100
pc2_ratio = pca.explained_variance_ratio_[1] * 100


# ---------------------------------------------------------
# 12. Plot the three clustering results
# ---------------------------------------------------------

fig, axes = plt.subplots(
    1,
    3,
    figsize=(20, 6)
)

for ax, (name, labels) in zip(
    axes,
    algorithms
):

    unique_labels = sorted(set(labels))

    # Treat DBSCAN noise separately
    normal_labels = [
        label for label in unique_labels
        if label != -1
    ]

    n_clusters = len(normal_labels)

    cmap = plt.get_cmap(
        "tab10",
        max(n_clusters, 1)
    )

    for i, label in enumerate(normal_labels):

        mask = labels == label

        ax.scatter(
            pc1[mask],
            pc2[mask],
            s=25,
            alpha=0.65,
            color=cmap(i),
            label=f"Cluster {label}"
        )

    # Plot DBSCAN noise in grey
    if -1 in unique_labels:

        noise = labels == -1

        ax.scatter(
            pc1[noise],
            pc2[noise],
            s=25,
            alpha=0.25,
            color="black",
            label="Noise"
        )

    ax.set_xlabel(
        f"PC1 ({pc1_ratio:.1f}%)"
    )

    ax.set_ylabel(
        f"PC2 ({pc2_ratio:.1f}%)"
    )

    ax.set_title(name)

    ax.legend(
        fontsize=8
    )


plt.tight_layout()
plt.show()
```
